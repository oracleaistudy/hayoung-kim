# Ch09-1 순차 데이터와 순환 신경망

## 1. 순차 데이터

`순차 데이터 Sequential data` 는 텍스트나, `시계열 데이터_time series data`와 같이 순서에 의미가 있는 데이터를 말함     


어떤 데이터가 먼저 들어가던지 상관 없던 것과 달리 순차데이터는 순서가 중요함         
예를들어, "별로지만 추천해요" 는 "별로지만"을 기억하여 완전히 긍정적인 글은 아니라고 판단.      

완전 연결 신경망이나, 합성곱 신경망은 이런 기억 장치가 없으므로, 정방향 계산을 수행하고 나면 그 샘플은 버려지고 다음 샘플을 처리할 때 재사용하지 않는다.        

### 피드포워드 신경망_Feedforward neural network(FFNN)

입력 데이터의 흐름이 앞으로만 전달되는 신경망 
- e.g. 완전 연결 신경망, 합성곱 신경망      


## 2. 순환 신경망_Recurrent neural network (RNN)

일반적인 완전 연결 신경망과 거의 비슷하나, 이전 데이터의 처리 결과가 순환되는 고리 하나가 추가된 것.        


뉴런의 출력이 다시 자기 자신으로 전달된다.      



Oa + B -> Ob (a, b)    
Ob + C -> Oc (a, b, c)      
...     

`타임스텝_timestep` 이라고 한다.        



일반적으로 은닉층의 활성화 함수로 하이퍼볼릭 탄젠트 함수인 `tanh`가 많이 사용됨     
(s 자 모양이라서 종종 시그모이드 함수라고도 불림)       

`tanh`함수는 -` ~ 1 사이의 범위     




## 3. 셀의 가중치와 입출력


순환층에 입력되는 특성의 개수가 4개이고 순환층의 뉴런이 3개라고 가정할 때,      
**Wx 의 크기는** 4 x 3 = 12개   (입력층과 순환층의 뉴런이 모두 완전 연결되기 때문)          
다음 타임스텝에 재사용되는 은닉 상태를 위한 가중치 **Wh의 크기는**      

3 x 3 = 9개     

**절편의 크기** 는 뉴런마다 하나의 절편         

**순환층의 파라미터 수** == Wx + Wh + 절편 = 12 + 9 + 3 = 24
